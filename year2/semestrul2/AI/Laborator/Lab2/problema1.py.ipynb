```python
import pandas as pd

def numar_si_tip_atribute(file_path):
    data = pd.read_csv(file_path , low_memory=False)
    num_atribute = len(data.columns)
    tip_atribute = data.dtypes.to_dict()

    print(f"Numărul de atribute: {num_atribute}")
    print("Tipurile de atribute:")
    for column, dtype in tip_atribute.items():
        print(f"{column}: {dtype}")

file_path = '/year2/semestrul2/AI/Laborator/Lab2/data/surveyDataSience.csv'
numar_si_tip_atribute(file_path)
```

```python
import pandas as pd

def calculate_average_study_years(file_path):
    data = pd.read_csv(file_path, low_memory=False)

    # Define the duration of each education level
    education_durations = {
        'bachelor': 3,
        'master': 2,
        'phd': 3
    }

    # Calculate total years of study for each respondent
    data['total_years'] = (data['bachelor'] * education_durations['bachelor'] +
                           data['master'] * education_durations['master'] +
                           data['phd'] * education_durations['phd'])

    # Calculate average duration for all respondents
    avg_all = data['total_years'].mean()

    # Calculate average duration for respondents from Romania
    romania_data = data[data['country'] == 'Romania']
    avg_romania = romania_data['total_years'].mean()

    # Calculate average duration for female respondents from Romania
    romania_female_data = romania_data[romania_data['gender'] == 'female']
    avg_romania_female = romania_female_data['total_years'].mean()

    print(f"Average duration of higher education years for all respondents: {avg_all:.2f} years")
    print(f"Average duration of higher education years for respondents from Romania: {avg_romania:.2f} years")
    print(f"Average duration of higher education years for female respondents from Romania: {avg_romania_female:.2f} years")

file_path = '/year2/semestrul2/AI/Laborator/Lab2/data/surveyDataSience.csv'
calculate_average_study_years(file_path)
```
