```python
import os
from PIL import Image
import numpy as np

def apply_sepia_filter(img):
    img = img.convert("RGB")
    np_img = np.array(img)

    sepia_matrix = np.array([
        [0.393, 0.769, 0.189],
        [0.349, 0.686, 0.168],
        [0.272, 0.534, 0.131]
    ])

    sepia_img = np_img @ sepia_matrix.T
    sepia_img = np.clip(sepia_img, 0, 255).astype(np.uint8)

    return Image.fromarray(sepia_img)

def convert_folder_to_sepia(input_folder):
    output_folder = input_folder + "_sepia"
    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):
            input_path = os.path.join(input_folder, filename)
            output_path = os.path.join(output_folder, filename)

            try:
                img = Image.open(input_path)
                sepia_img = apply_sepia_filter(img)
                sepia_img.save(output_path)
                print(f"Converted: {filename}")
            except Exception as e:
                print(f"❌ Failed to convert {filename}: {e}")

    print(f"\n✅ All images from '{input_folder}' were saved in '{output_folder}'.")

# Convert both folders
convert_folder_to_sepia("dataset_cow")
convert_folder_to_sepia("dataset_lion")
```
